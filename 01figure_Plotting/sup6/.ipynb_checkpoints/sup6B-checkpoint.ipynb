```r
pdf("gene_msmc_raw_layer.pdf",width = 5.5,height = 4)
for(gene in gene1){
    s=0
    for(i in week){
        s=s+1
        w <- subset(mer,subset = week==i)
        fib <- subset(w,subset = first_type=="Fibroblast")
        msmc <- subset(w,subset = first_type!="Fibroblast")
        if(s==1){
            m <- msmc
        }
        else{
            m <- rbind(m,msmc)
        }
    }
    m$week <- factor(m$week,level=week)
    p <- ggplot(m,aes(x=layer1,y = get(gene),fill=week,color=week))+theme_classic ()+
    labs(subtitle = gene)+geom_smooth(method = "gam",span=1,se = T,level=0.90,alpha=0.15)+scale_fill_manual(values = cell_type_cols)+scale_color_manual(values = cell_type_cols)+
    labs(x="digitization_layer",y="expression")
    print(p)
}
dev.off()
```

```r
pdf("gene_msmc_fib_mean_layer.pdf",width = 5.5,height = 4)
for(gene in gene1){
    s=0
    for(i in order){
        s=s+1
        w <- subset(mer,subset = week==i)
        fib <- subset(w,subset = first_type=="Fibroblast")
        msmc <- subset(w,subset = first_type!="Fibroblast")
        msmc[,gene] <- msmc[,gene]/mean(fib[,gene])
        if(s==1){
            m <- msmc
        }
        else{
            m <- rbind(m,msmc)
        }
    }
    m$week <- factor(m$week,level=order)
    p <- ggplot(m,aes(x=layer1,y = get(gene),fill=week,color=week))+theme_classic ()+
    labs(subtitle = gene)+geom_smooth(method = "gam",span=1,se = T,level=0.90,alpha=0.15)+scale_fill_manual(values = cell_type_cols)+scale_color_manual(values = cell_type_cols)+
    labs(x="digitization_layer",y="expression")
    print(p)
}
dev.off()
```
