```r
stage <- c('stage2','stage3','stage4')
cell <- c('Fibroblast','SMC','Modulated_SMC','Macrophage','EC')
n=0
for(i in stage){
    n=n+1
    s <- subset(mer2,subset = satge_p_all%in%c("stage1",i))
    c <- subset(s,subset = first_type=="Macrophage")
    Idents(c) <- as.factor(c@meta.data$satge_p_all)
    deg = FindAllMarkers(c,only.pos = FALSE,
                               min.pct = 0.06,
                               logfc.threshold = 0.5)
    deg <- subset(deg,subset=p_val_adj<0.05)
    d <- subset(deg,subset = cluster=="stage1")
    d <- subset(d,select = c("gene"))
    d[,1] <- 1
    colnames(d) <- i
    if(n==1){
        mer <- d
    }
    else{
        mer <- merge(mer,d,all = TRUE,by = 0)
        rownames(mer) <- mer$Row.names
        mer <- mer[,-1]
        
    }
        
}
mer[is.na(mer)] <- 0
week <- c('stage4','stage3','stage2')
upset(mer, week,min_degree = 1,name = "Macrophage",sort_sets=FALSE)
ggsave("MP_upset.pdf",width = 6,height = 5)
```

```r
# 创建棒棒糖图
p <- ggplot(data, aes(x = Count, y = Description)) +
  # 添加点
  geom_point(aes(color = Description), size = 8) +
  # 添加点的外圈
  geom_point(aes(color = Description), size = 10, shape = 21, fill = NA) +
  # 添加线条
  geom_segment(aes(x = 0, xend = Count, y = Description, yend = Description, color = Description), linewidth = 1) +
  # 设定颜色
  scale_color_manual(values = Color) +
  # 设置主题
  theme_classic() +
  theme(legend.position = "none",  # 移除图例
        panel.border = element_rect(color = "black", fill = NA, linewidth = 1), # 添加边框
        axis.line = element_line(color = "black")) + # 添加次网格线
  # 设置x轴范围
  xlim(0, len) +
  # 设置坐标轴标签
  labs(x = "gene Ratio", y = "") +geom_text(aes(label = p.adjust), hjust = -0.4)+theme(axis.title.x=element_text(vjust=0, size=20,face = "plain"))+#调整xlab字体大小
    theme(axis.title.y=element_text(vjust=0, size=20,face = "plain"))+#调整ylab字体大小
    theme(plot.title = element_text(size = 20, face = "bold"))+
    theme(axis.text.y=element_text(vjust=0, size=15,face = "plain"))+
    theme(axis.line = element_line(color = "black",linewidth = 0.8))
print(p)
```
